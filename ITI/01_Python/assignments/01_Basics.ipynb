```python
# 3) The volume of a sphere with radius r is 4/3πr3.
import math

r = 5 # radius in centimeters (cm)
volume = (4 / 3) * math.pi * (r ** 3) # volume in cubic centimeters (cm^3)
print("Volume =", volume)
```

```python
# 5) Check if a Number is Even or Odd
num = int(input("Enter a number: "))
if num % 2 == 0:
    print(num, "is even")
else:
    print(num, "is odd")
```

```python
# 6) Find the Largest of Three Numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))
print("Max number =", max(num1, num2, num3))
```

```python
# 7) Temperature Converter
C = float(input("Enter a temperature degree in Celsius: "))
F = (9 / 5) * C + 32
print("Temperature in Fahrenheit =", F)
```
