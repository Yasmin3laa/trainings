## 3. Sphere Volume

Calculate the volume of a sphere using the following formula:

$$
V = \frac{4}{3}\pi r^3
$$

Where `r` represents the radius of the sphere.

```python
import math

radius = 5  # radius in centimeters (cm)
volume = (4 / 3) * math.pi * (radius ** 3)

print("Volume =", volume, "cm³")
```

## 5. Even or Odd Number

Write a program that takes a number from the user and checks whether it is even or odd.

```python
number = int(input("Enter a number: "))

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
```

## 6. Largest of Three Numbers

Write a program that takes three numbers from the user and finds the largest number.

```python
number_1 = float(input("Enter first number: "))
number_2 = float(input("Enter second number: "))
number_3 = float(input("Enter third number: "))

largest_number = max(number_1, number_2, number_3)

print("Largest number =", largest_number)
```

## 7. Temperature Converter

Write a program that converts a temperature from Celsius to Fahrenheit using the following formula:

$$
F = \frac{9}{5}C + 32
$$

```python
celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (9 / 5) * celsius + 32

print("Temperature in Fahrenheit =", fahrenheit)
```

## 8. Simple Interest Calculator

Calculate the Simple Interest (SI) using the following formula:

$$
SI = \frac{P \times R \times T}{100}
$$

Where:

- `P` = Principal amount
- `R` = Rate of interest
- `T` = Time period

```python
principal = 6
rate = 7
time = 8

simple_interest = (principal * rate * time) / 100

print("Simple Interest (SI) =", simple_interest)
```

## 9. Days to Seconds Converter

Write a program that takes the number of days from the user and converts it into seconds.

```python
days = int(input("How many days? "))

seconds = days * 24 * 60 * 60

print("Seconds in that many days:", seconds)
```
