# Strings and Lists in python

```python
# 1- Fill an array of 5 elements from the user, Sort it in descending and ascending orders then display the output
array = []

for num in range(5):
    array.append(int(input(f"Enter number {num + 1}: ")))

array.sort()
print(f"ASC: {array}")

array.sort(reverse=True)
print(f"DESC: {array}")
```

```python
# 2- Write a program that prints the number of times the string 'iti' occurs in anystring.
string = "Python Track, Information Technology Institute iti"
print(string.count("iti"))
```

```python
# 3- Write a program that remove all vowels from the input word and generate a brief version of it
vowels = "aeiouAEIOU"

word = input("Enter a word: ")
new_word = ""

for char in word:
    if char not in vowels:
        new_word += char

print(new_word)
```

```python
# 4- Write a program that counts up the number of vowels [a, e, i, o, u] contained in the string.
vowels = "aeiouAEIOU"
string = "Python Track, Information Technology Institute iti"
vowel_count = 0

for char in string:
    if char in vowels:
        vowel_count += 1

print(vowel_count)
```

```python
# 5- Create a list of integers from 1 to 10. Write a script to:
# - Print the first and last elements.
# - Reverse the list.
# - Find the sum of all elements.
my_list = [1, 10, 3, 8, 5, 6, 2, 9, 4, 7]

print(f"First element = {my_list[0]}, Last element = {my_list[-1]}")

my_list.reverse()
print(f"Reversed list = {my_list}")

total_sum = 0
for num in my_list:
    total_sum += num
print(f"Sum = {total_sum}")
```

# Sets in python

```python
# 1- Create a set of the first five even numbers. Write a script to:
# - Add a new number to the set.
# - Remove a number from the set.
# - Check if a specific number exists in the set.
# - Perform a union and intersection with another set {2, 4, 6, 8, 10}.
my_set = set(range(0, 10, 2))
print(my_set)

my_set.add(10)
print(my_set)

print(my_set.pop())
print(my_set)

print(12 in my_set)

another_set = {2, 4, 6, 8, 10}
print(my_set.union(another_set))
print(my_set.intersection(another_set))
```

```python
# 2- Set-Based Password Security Checker
def check_password(password):
    password_set = {"123456", "password", "123456789", "12345", "12345678", "qwerty", "1234567", "111111", "123123", "abc123", "password1", "1234"}
    if password in password_set:
        print("Your password is weak")
    else:
        print("Your password is strong")

password = input("Enter your password: ")
check_password(password)
```

# Tuples in Python

```python
# Task 1: Working with Tuples
# Create a tuple of five different fruits. Write a script to:
# - Print the first and last fruit.
# - Slice the tuple to get the first three fruits.
# - Count the occurrences of a specific fruit.
# - Try modifying the tuple and handle the error gracefully.
my_tuple = ("banana", "apple", "strawberry", "mango", "apple")

print(f"First element: {my_tuple[0]}, Last element: {my_tuple[-1]}")
print(f"First 3 fruits: {my_tuple[:3]}")
print(f"Count apple = {my_tuple.count('apple')}")

# Handling tuple modification error gracefully using try-except
try:
    my_tuple[4] = "cherry"
except TypeError as e:
    print(f"Error handling: Tuples are immutable and cannot be modified! Details: {e}")

# Modifying via list conversion
my_tuple = list(my_tuple)
my_tuple[4] = "cherry"
my_tuple = tuple(my_tuple)
print(f"Last element after converting & modifying: {my_tuple[4]}")
```
