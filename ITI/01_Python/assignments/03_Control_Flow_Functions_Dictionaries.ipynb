# Python Tasks & Dictionaries

```python
# 1) Calculator
num1 = int(input("Number 1: "))
num2 = int(input("Number 2: "))
operation = input("Operation: ")

operations = {
    "add": ['+', num1 + num2],
    "subtract": ['-', num1 - num2],
    "multiply": ['*', num1 * num2]
}

if operation in operations:
    print(f"{num1} {operations[operation][0]} {num2} = {operations[operation][1]}")
```

```python
# 2) FizzBuzz
num = int(input("Number: "))

if num % 3 == 0 and num % 5 == 0:
    print("FizzBuzz")
elif num % 3 == 0:
    print("Fizz")
elif num % 5 == 0:
    print("Buzz")
```

```python
# 3) PIN and number of attempts
pin = input("PIN: ")
tries = 1

if pin == "4321":
    print("Correct! It only took you one single attempt!")
else:
    while pin != "4321":
        print("Wrong")
        pin = input("PIN: ")
        tries += 1
    print(f"Correct! It took you {tries} attempts")
```

```python
# 4) Function to find the greatest number without max()
def greatest_number1(num1, num2, num3):
    max_num = num1
    if num2 >= max_num:
        max_num = num2
    if num3 >= max_num:
        max_num = num3
    return max_num

def greatest_number2(num1, num2, num3):
    my_list = [num1, num2, num3]
    my_list.sort()
    return my_list[-1]

def greatest_number3(*numbers):
    max_num = numbers[0]
    for num in numbers:
        if num > max_num:
            max_num = num
    return max_num

print(f"First way => max = {greatest_number1(5, 0, 10)}")
print(f"Second way => max = {greatest_number2(5, 0, 10)}")
print(f"Third way => max = {greatest_number3(5, 0, 10)}")
```

# Dictionaries

```python
# 1. Squares numbers
my_dict = {
    1: 1,
    2: 4,
    3: 9,
    4: 16,
    5: 25
}

print("Initial Dictionary:")
for key, value in my_dict.items():
    print(f"{key}: {value}")

# Adding key-value pair 6: 36
my_dict[6] = 36

print("\nAfter Adding 6: 36:")
for key, value in my_dict.items():
    print(f"{key}: {value}")

print(f"\nIs 4 in the dictionary? {4 in my_dict}")
```

```python
# 2. Student Grades
students = {
    "Alice": [85, 90, 78],
    "Bob": [92, 88, 95],
    "Charlie": [70, 75, 80],
    "David": [95, 98, 100]
}

average_grades = {}

for key, value in students.items():
    average_grades[key] = round((sum(value) / len(value)), 2)
    print(f"{key}'s average grade = {average_grades[key]}")

max_average = max(average_grades.values())
max_average_student = ""

for key, value in average_grades.items():
    if value == max_average:
        max_average_student = key

print(f"\nThe student with the highest average is \"{max_average_student}\" with an average grade = {max_average}")
```
