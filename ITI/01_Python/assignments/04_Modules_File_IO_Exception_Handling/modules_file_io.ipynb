# 📁 File I/O

## Task 1 — User Information

Create a program that collects user information and stores it in `users.txt`.

### Requirements

The program should:

1. Ask the user for:
   - Name
   - Age
   - Email
   - Track
2. Validate the entered data.
3. Write the information to `users.txt`.
4. Read the file contents.
5. Display the contents in uppercase.
6. Count the total number of words.

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
email = input("Enter your email: ")
track = input("Enter your track: ")

if age < 20:
    raise Exception("Age must be greater than 20.")

if not isinstance(age, int):
    raise ValueError("Age must be an integer")

for char in name:
    if char.isalpha() or char.isspace():
        continue
    else:
        print("Name must only contain alphabetic characters and spaces.")
        break

print(f"Name: {name}")

for char in track:
    if char.isalpha() or char.isspace():
        continue
    else:
        print("Track must only contain alphabetic characters and spaces.")
        break

print(f"Track: {track}")

if age < 1:
    print("Age must be a positive integer.")

print(f"Age: {age}")

if '@' not in email or '.' not in email:
    print("Email must contain '@' and '.'")

print(f"Email: {email}")

try:
    with open("users.txt", "w") as file:
        file.write(
            f"Name: {name}, Age: {age}, "
            f"Email: {email}, Track: {track}\n"
        )

    with open("users.txt", "r") as file:
        content = file.readlines()

    total_words = 0

    for line in content:
        print(line.upper(), end="")
        total_words += len(line.split())

    print(f"Total words = {total_words}")

except Exception as error:
    print(f"Something went wrong: {error}")
```

## Task 2 — Reading a File

Create a program that asks the user for a file name and displays its contents.

### Requirements

- Ask the user for a file name.
- Attempt to open and read the file.
- Handle `FileNotFoundError`.
- Allow the user to try another file if the requested file does not exist.
- Stop when the file is successfully opened or the user chooses not to continue.

```python
file_name = input("Enter file name: ")
try_again = "yes"

while try_again.lower() == "yes":
    try:
        with open(file_name, "r") as file:
            print(f"Contents of {file_name}:")
            print(file.read())
            break

    except FileNotFoundError:
        print(f"Error: File '{file_name}' not found.")

        try_again = input(
            "Would you like to try again? (yes/no): "
        )

        if try_again.lower() == "yes":
            file_name = input("Enter file name: ")

        elif try_again.lower() != "no":
            print("Invalid answer, only yes or no.")
            break
```
