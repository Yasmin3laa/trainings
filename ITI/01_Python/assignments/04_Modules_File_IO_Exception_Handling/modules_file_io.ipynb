## Modules

### Task 1: Module Creation and Import Testing

```python
# Creating math_utils.py file programmatically
math_utils_code = '''\
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def division(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "ZeroDivisionError: Cannot divide by zero!"
'''

with open("math_utils.py", "w") as f:
    f.write(math_utils_code)

print("math_utils.py created successfully!")
```

```python
# 1) Import the module and test all functions
import math_utils

print("--- Standard Import ---")
print("Add:", math_utils.add(1, 2))
print("Subtract:", math_utils.subtract(4, 3))
print("Multiply:", math_utils.multiply(5, 6))
print("Division:", math_utils.division(8, 4))
print("Division by zero:", math_utils.division(8, 0))

# 2) Import math_utils as mu
import math_utils as mu

print("\n--- Import as mu ---")
print("Add:", mu.add(10, 5))
print("Subtract:", mu.subtract(10, 5))
print("Multiply:", mu.multiply(10, 5))
print("Division:", mu.division(10, 5))

# 3) Import only add from math_utils
from math_utils import add

print("\n--- Import only add ---")
print("Call without module name:", add(100, 200))
```

## File I/O

### Task 1: User Input Validation and File Handling

```python
try:
    name = input("Enter your name: ")
    age_input = input("Enter your age: ")
    email = input("Enter your email: ")
    track = input("Enter your track: ")

    # Validate Age as integer
    try:
        age = int(age_input)
    except ValueError:
        raise ValueError("Age must be an integer!")

    # Check if age is less than 20
    if age < 20:
        raise Exception("Age must be at least 20.")

    # Validate Name & Track (Only alphabetic characters and spaces)
    if not all(char.isalpha() or char.isspace() for char in name) or not name.strip():
        raise ValueError("Name must only contain alphabetic characters and spaces.")

    if not all(char.isalpha() or char.isspace() for char in track) or not track.strip():
        raise ValueError("Track must only contain alphabetic characters and spaces.")

    # Validate Email Format (must contain '@' and '.')
    if "@" not in email or "." not in email:
        raise ValueError("Email must follow a standard format containing '@' and '.'.")

    # Write data to users.txt
    user_line = f"Name: {name}, Age: {age}, Email: {email}, Track: {track}\n"
    with open("users.txt", "w") as file:
        file.write(user_line)
    print("\nData saved successfully to users.txt!\n")

    # Read content from users.txt and print each line in uppercase
    print("--- File Content (Uppercase) ---")
    with open("users.txt", "r") as file:
        content = file.readlines()
        total_words = 0
        for line in content:
            print(line.upper().strip())
            words = line.split()
            total_words += len(words)
            
        print(f"\nTotal word count = {total_words}")

except Exception as e:
    print(f"Custom Error: {e}")
```

### Task 2: File Reader with Retry Mechanism

```python
while True:
    file_name = input("Enter file name: ")
    try:
        with open(file_name, "r") as file:
            print(f"\nContents of {file_name}:")
            print(file.read())
            break
    except FileNotFoundError:
        print(f"Error: File '{file_name}' not found.")
        try_again = input("Would you like to try again? (yes/no): ").strip().lower()
        if try_again != "yes":
            print("Exiting program.")
            break
```
