## Modules

### Task 1: Module Creation and Import Testing

```python
# Creating math_utils.py file programmatically
math_utils_code = '''
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def division(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero!"
'''

with open("math_utils.py", "w") as f:
    f.write(math_utils_code)

print("math_utils.py created successfully!")
```

```python
# 1. Basic import and testing
import math_utils

print("--- Standard Import ---")
print("Add:", math_utils.add(1, 2))
print("Subtract:", math_utils.subtract(4, 3))
print("Multiply:", math_utils.multiply(5, 6))
print("Division:", math_utils.division(8, 4))
print("Division by Zero:", math_utils.division(8, 0))

# 2. Import with alias 'mu'
import math_utils as mu

print("\n--- Alias Import (mu) ---")
print("Add:", mu.add(10, 5))
print("Subtract:", mu.subtract(10, 5))
print("Multiply:", mu.multiply(10, 5))
print("Division:", mu.division(10, 5))

# 3. Importing specific function without module name
from math_utils import add

print("\n--- Direct Import ---")
print("Add directly:", add(100, 200))
```

## File I/O

### Task 1: User Validation and File Operations

```python
try:
    name = input("Enter your name: ")
    track = input("Enter your track: ")
    email = input("Enter your email: ")
    age_input = input("Enter your age: ")

    # Validate Age as integer
    if not age_input.isdigit():
        raise ValueError("Age must be a positive integer.")
    
    age = int(age_input)

    # Check age range exception
    if age < 20:
        raise Exception("Age must be at least 20.")

    # Validate Name & Track
    if not all(char.isalpha() or char.isspace() for char in name):
        raise ValueError("Name must only contain alphabetic characters and spaces.")
    
    if not all(char.isalpha() or char.isspace() for char in track):
        raise ValueError("Track must only contain alphabetic characters and spaces.")

    # Validate Email Format
    if "@" not in email or "." not in email:
        raise ValueError("Email must follow a standard email format containing '@' and '.'.")

    # Format and Write to users.txt
    user_data = f"Name: {name}, Age: {age}, Email: {email}, Track: {track}\n"
    with open("users.txt", "w") as file:
        file.write(user_data)
    print("Data written to users.txt successfully!\n")

    # Read from users.txt, print uppercase, and count total words
    with open("users.txt", "r") as file:
        content = file.readlines()
        total_words = 0
        print("--- File Content (In Uppercase) ---")
        for line in content:
            print(line.upper().strip())
            words = line.split()
            total_words += len(words)
            
        print(f"\nTotal words count: {total_words}")

except ValueError as ve:
    print(f"Validation Error: {ve}")
except Exception as e:
    print(f"Custom Error Message: {e}")
```

### Task 2: File Reader with Retry Logic

```python
while True:
    file_name = input("Enter file name: ")
    try:
        with open(file_name, "r") as file:
            print(f"\nContents of {file_name}:")
            print(file.read())
            break
    except FileNotFoundError:
        print(f"Error: File '{file_name}' not found.")
        try_again = input("Would you like to try again? (yes/no): ").strip().lower()
        if try_again != "yes":
            print("Program exited.")
            break
```
