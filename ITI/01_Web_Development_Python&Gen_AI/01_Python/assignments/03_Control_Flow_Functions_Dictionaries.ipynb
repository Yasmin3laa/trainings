# 🔀 Control Flow & Functions

## 1. Calculator

Create a simple calculator that accepts two numbers and an operation, then displays the calculated result.

```python
num1 = int(input("Number 1: "))
num2 = int(input("Number 2: "))
operation = input("Operation (+, -, *, /): ")

operations = {
    "+": num1 + num2,
    "-": num1 - num2,
    "*": num1 * num2
}

if operation == "/":
    if num2 != 0:
        result = num1 / num2
        print(f"{num1} / {num2} = {result}")
    else:
        print("Cannot divide by zero.")
elif operation in operations:
    print(f"{num1} {operation} {num2} = {operations[operation]}")
else:
    print("Invalid operation.")
```

## 2. FizzBuzz

Write a program that:

- Prints `FizzBuzz` if the number is divisible by both 3 and 5.
- Prints `Fizz` if it is divisible by 3.
- Prints `Buzz` if it is divisible by 5.
- Prints the number otherwise.

```python
num = int(input("Number: "))

if num % 3 == 0 and num % 5 == 0:
    print("FizzBuzz")
elif num % 3 == 0:
    print("Fizz")
elif num % 5 == 0:
    print("Buzz")
else:
    print(num)
```

## 3. PIN Validation and Attempt Counter

Write a program that asks the user to enter a PIN and continues asking until the correct PIN is entered. The program should also count the number of attempts.

```python
correct_pin = "4321"
pin = input("PIN: ")
attempts = 1

while pin != correct_pin:
    print("Wrong PIN")
    pin = input("PIN: ")
    attempts += 1

if attempts == 1:
    print("Correct! It only took you one attempt!")
else:
    print(f"Correct! It took you {attempts} attempts.")
```

## 4. Find the Greatest Number

Create functions that find the greatest number among three or more numbers without directly using the built-in `max()` function.

```python
def greatest_number_1(num1, num2, num3):
    greatest = num1

    if num2 >= greatest:
        greatest = num2

    if num3 >= greatest:
        greatest = num3

    return greatest


def greatest_number_2(num1, num2, num3):
    numbers = [num1, num2, num3]
    numbers.sort()
    return numbers[-1]


def greatest_number_3(*numbers):
    greatest = numbers[0]

    for number in numbers:
        if number > greatest:
            greatest = number

    return greatest


print(f"First way  => greatest = {greatest_number_1(5, 0, 10)}")
print(f"Second way => greatest = {greatest_number_2(5, 0, 10)}")
print(f"Third way  => greatest = {greatest_number_3(5, 0, 10)}")
```

# 📚 Dictionaries

## 5. Squares Dictionary

Create a dictionary where each number from 1 to 5 is mapped to its square. Then add the key-value pair `6: 36` and check whether a specific key exists.

```python
squares = {
    1: 1,
    2: 4,
    3: 9,
    4: 16,
    5: 25
}

print("Initial Dictionary:")
for key, value in squares.items():
    print(f"{key}: {value}")

squares[6] = 36

print("\nAfter Adding 6: 36:")
for key, value in squares.items():
    print(f"{key}: {value}")

print(f"\nIs 4 in the dictionary? {4 in squares}")
```

## 6. Student Grades

Given a dictionary containing students and their grades:

- Calculate each student's average grade.
- Store the averages in a new dictionary.
- Find the student with the highest average.

```python
students = {
    "Alice": [85, 90, 78],
    "Bob": [92, 88, 95],
    "Charlie": [70, 75, 80],
    "David": [95, 98, 100]
}

average_grades = {}

for student, grades in students.items():
    average_grades[student] = round(sum(grades) / len(grades), 2)
    print(f"{student}'s average grade = {average_grades[student]}")

highest_average = max(average_grades.values())
highest_average_student = ""

for student, average in average_grades.items():
    if average == highest_average:
        highest_average_student = student

print(
    f"\nThe student with the highest average is "
    f"{highest_average_student} with an average grade = {highest_average}"
)
```
