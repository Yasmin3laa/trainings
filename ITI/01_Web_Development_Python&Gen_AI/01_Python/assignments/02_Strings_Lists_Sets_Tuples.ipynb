# 📌 Strings & Lists

## 1. Sort a List

Fill a list with five numbers provided by the user, then display the list in ascending and descending order.

```python
numbers = []

for index in range(5):
    numbers.append(int(input(f"Enter number {index + 1}: ")))

numbers.sort()
print(f"Ascending: {numbers}")

numbers.sort(reverse=True)
print(f"Descending: {numbers}")
```

## 2. Count String Occurrences

Write a program that counts how many times the string `iti` occurs in a given string.

```python
text = "Python Track, Information Technology Institute iti"

occurrences = text.count("iti")

print("Occurrences of 'iti':", occurrences)
```

## 3. Remove Vowels from a String

Write a program that removes all vowels from an input word and generates a shortened version of it.

```python
vowels = "aeiouAEIOU"

word = input("Enter a word: ")
shortened_word = ""

for character in word:
    if character not in vowels:
        shortened_word += character

print("Shortened word:", shortened_word)
```

## 4. Count Vowels

Write a program that counts the number of vowels (`a`, `e`, `i`, `o`, `u`) contained in a string.

```python
vowels = "aeiouAEIOU"
text = "Python Track, Information Technology Institute iti"
vowel_count = 0

for character in text:
    if character in vowels:
        vowel_count += 1

print("Number of vowels:", vowel_count)
```

## 5. Basic List Operations

Create a list of integers from 1 to 10 and:

- Print the first and last elements.
- Reverse the list.
- Calculate the sum of all elements.

```python
numbers = [1, 10, 3, 8, 5, 6, 2, 9, 4, 7]

print(f"First element = {numbers[0]}")
print(f"Last element = {numbers[-1]}")

numbers.reverse()
print(f"Reversed list = {numbers}")

total_sum = 0

for number in numbers:
    total_sum += number

print(f"Sum = {total_sum}")
```

# 🔢 Sets

## 6. Set Operations

Create a set containing the first five even numbers. Then:

- Add a new number to the set.
- Remove a specific number from the set.
- Check whether a specific number exists in the set.
- Perform a union with another set.
- Perform an intersection with another set.

```python
numbers_set = set(range(0, 10, 2))
print("Original set:", numbers_set)

numbers_set.add(10)
print("After adding 10:", numbers_set)

numbers_set.remove(0)
print("After removing 0:", numbers_set)

print("Is 12 in the set?", 12 in numbers_set)

another_set = {2, 4, 6, 8, 10}

print("Union:", numbers_set.union(another_set))
print("Intersection:", numbers_set.intersection(another_set))
```

## 7. Set-Based Password Security Checker

Use a set containing commonly used passwords to check whether a user's password appears in the list of weak passwords.

```python
def check_password(password):
    weak_passwords = {
        "123456",
        "password",
        "123456789",
        "12345",
        "12345678",
        "qwerty",
        "1234567",
        "111111",
        "123123",
        "abc123",
        "password1",
        "1234"
    }

    if password in weak_passwords:
        print("Your password is weak")
    else:
        print("Your password is not in the weak password list")


password = input("Enter your password: ")
check_password(password)
```

# 📦 Tuples

## 8. Working with Tuples

Create a tuple containing five fruits and:

- Print the first and last fruits.
- Slice the tuple to get the first three fruits.
- Count the occurrences of a specific fruit.
- Try modifying the tuple and handle the resulting error.
- Convert the tuple to a list to modify an element, then convert it back to a tuple.

```python
fruits = ("banana", "apple", "strawberry", "mango", "apple")

print(f"First fruit: {fruits[0]}")
print(f"Last fruit: {fruits[-1]}")
print(f"First 3 fruits: {fruits[:3]}")
print(f"Count of apple: {fruits.count('apple')}")

# Tuples are immutable and cannot be modified directly.
try:
    fruits[4] = "cherry"
except TypeError as error:
    print(f"Error: Tuples are immutable and cannot be modified. Details: {error}")

# Convert the tuple to a list to modify an element.
fruits_list = list(fruits)
fruits_list[4] = "cherry"

# Convert the list back to a tuple.
fruits = tuple(fruits_list)

print(f"Last fruit after modification: {fruits[4]}")
```
